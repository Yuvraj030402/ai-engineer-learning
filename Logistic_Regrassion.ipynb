```python
import pandas as pd

t = pd.read_csv("https://raw.githubusercontent.com/hakimyameen/Datasets/refs/heads/main/PimaIndiansDiabetes.csv")
t
```

```python
#X  = t.iloc[:,:-1]
#X = t.drop(columns='Class')
#X = t.loc[:,'TimesPregnant':'Age']
X = t[['TimesPregnant', 'GlucoseConcentration', 'BloodPrs', 'SkinThickness',
       'Serum', 'BMI', 'DiabetesFunct', 'Age']]
X[:1]
```

```python
y = t['Class']
y[:1]
```

```python
y.value_counts()
```

```python
y.value_counts(normalize=True) * 100
```

```python
from sklearn.model_selection import train_test_split
X_train , X_test , y_train , y_test  = train_test_split(X,y,train_size=.8,random_state=42)
```

```python
# laod algo
import warnings   # to ignore warnings we use this
warnings.filterwarnings('ignore')
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()
model.fit(X_train,y_train)
```

```python
model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score , r2_score
y_pred = model.predict(X_test)
print(accuracy_score(y_test,y_pred))
```

### Create and Launch Gradio UI
This interface captures the input features, runs the trained `model`, displays the prediction probability as an attractive visual indicator, and summarizes the user's selected values.

```python
import gradio as gr
import numpy as np

def predict_diabetes(pregnancies, glucose, bp, skin, insulin, bmi, pedigree, age):
    # Prepare the input vector
    input_data = np.array([[pregnancies, glucose, bp, skin, insulin, bmi, pedigree, age]])

    # Get class prediction and probabilities
    prediction = model.predict(input_data)[0]
    probabilities = model.predict_proba(input_data)[0]
    prob_diabetes = probabilities[1] * 100
    prob_healthy = probabilities[0] * 100

    # Style and format the result message
    if prediction == 1:
         status = "⚠️ High Risk of Diabetes"
         score_highlight = f"<div style='background-color: #ffe6e6; border-left: 6px solid #ff4d4d; padding: 15px; border-radius: 5px;'><b style='color: #cc0000; font-size: 20px;'>{status}</b><br/>Confidence Score: <b>{prob_diabetes:.2f}%</b> probability of diabetes.</div>"
    else:
         status = "✅ Low Risk (Healthy)"
         score_highlight = f"<div style='background-color: #e6ffe6; border-left: 6px solid #33cc33; padding: 15px; border-radius: 5px;'><b style='color: #006600; font-size: 20px;'>{status}</b><br/>Confidence Score: <b>{prob_healthy:.2f}%</b> probability of being healthy.</div>"

    # Create a summary of selected values
    summary = f"""
    ### Selected Inputs Summary:
    * **Pregnancies**: {pregnancies}
    * **Glucose**: {glucose} mg/dL
    * **Blood Pressure**: {bp} mmHg
    * **Skin Thickness**: {skin} mm
    * **Insulin (Serum)**: {insulin} μU/mL
    * **BMI**: {bmi}
    * **Pedigree Function**: {pedigree}
    * **Age**: {age} years
    """
    return score_highlight, summary

# Design an attractive Gradio blocks interface
with gr.Blocks(theme=gr.themes.Soft()) as demo:
    gr.Markdown("# 🩸 Pima Indians Diabetes Predictor")
    gr.Markdown("Input patient metrics below to evaluate the likelihood of diabetes.")

    with gr.Row():
        with gr.Column():
            gr.Markdown("### Patient Demographics & Vitals")
            pregnancies = gr.Slider(0, 20, value=3, step=1, label="Times Pregnant")
            age = gr.Slider(10, 100, value=33, step=1, label="Age (years)")
            bmi = gr.Slider(10.0, 70.0, value=32.0, step=0.1, label="BMI (Body Mass Index)")
            pedigree = gr.Slider(0.0, 3.0, value=0.5, step=0.01, label="Diabetes Pedigree Function")

        with gr.Column():
            gr.Markdown("### Clinical Lab Measurements")
            glucose = gr.Slider(0, 200, value=120, step=1, label="Glucose Concentration (2-hour OGTT)")
            bp = gr.Slider(0, 150, value=70, step=1, label="Diastolic Blood Pressure (mmHg)")
            skin = gr.Slider(0, 100, value=20, step=1, label="Triceps Skin Fold Thickness (mm)")
            insulin = gr.Slider(0, 900, value=80, step=1, label="2-Hour Serum Insulin (μU/mL)")

    btn = gr.Button("🔮 Predict Diabetes Risk", variant="primary")

    gr.Markdown("### Evaluation Results")
    output_score = gr.HTML(label="Prediction Outcome")
    output_summary = gr.Markdown(label="Selected Values Details")

    btn.click(
        fn=predict_diabetes,
        inputs=[pregnancies, glucose, bp, skin, insulin, bmi, pedigree, age],
        outputs=[output_score, output_summary]
    )

demo.launch(debug=True, share=True)
```

### Step to Improve Accuracy: Feature Scaling
Logistic Regression is highly sensitive to the scale of input features. Since features like `Serum` (Insulin) and `GlucoseConcentration` have vastly larger numerical ranges than `DiabetesFunct` or `TimesPregnant`, applying standard scaling is a critical step to boost the model's convergence and general accuracy.

```python
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Create a pipeline that scales the input data and then trains the logistic regression
improved_model = make_pipeline(StandardScaler(), LogisticRegression(random_state=42))
improved_model.fit(X_train, y_train)

# Test accuracy
y_pred_scaled = improved_model.predict(X_test)
print(f"Baseline Accuracy: {accuracy_score(y_test, y_pred):.4f}")
print(f"Scaled Model Accuracy: {accuracy_score(y_test, y_pred_scaled):.4f}")
```
