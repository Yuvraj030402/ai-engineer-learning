```python
# it is simple way to create list in one line code
```

```python
[i**2 for i in range(1,5) if i % 2 ==0]
```

**Ternary Operator**

```python
#A ternary operator is a short way to write an if-else condition in a single line
```

```python
age = 40

print("Eligible" if age >= 18 else "Not Eligible")
```

```python
#result = "A" if marks >= 90 else ("B" if marks >= 75 else ("C" if marks >= 50 else "Fail"))
marks = 85

result = "A" if marks >= 90 else "B" if marks >= 75 else "C" if marks >= 50 else "Fail"

print(result)
```

list comprehension with a ternary **operator** **bold text**

```python
marks = [25,35,95,43,76,77]

result = ["A" if mark >= 90 else "B" if mark >= 75 else "C" if mark >= 50 else "Fail" for mark in marks]


print(result)
```
