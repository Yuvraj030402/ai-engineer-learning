**Lambda function**

syntax: lambda parameter/s:expression

we can supply multiple parameters/single parameter

Only one expression is allowed

lambda is a python keyword

it has default return

It is also known as Anonymous/Nameless function

```python
test = lambda num: '+ve' if num > 0 else '-ve'
test(3)
test(-3)
```

```python
test = lambda age , validAge : '+ve' if age > validAge else '-ve'
test(30,17)
```

**Higher Order function**

map(function,iterable)

filter(function,iterable)

reduce(function,iterable)


```python
#1 map(func,iterable) - operator base oprations
s = [10,20,30]
test = lambda num: num+1
list(map(test,s))
```

```python
s = [10,20,30]

def test(a):
    return a+1
list(map(test,s))
```

```python
#2 filter(function,iterable) - cond base oprator
    #means if condition is True only those elements will be fetched
s = [10,30,24]
eligible  = lambda age : age > 18
list(filter(eligible,s))
```

```python
s = [10,30,24]
def eligible(age):
  if age > 18:
     return 1
list(filter(eligible,s))
```
