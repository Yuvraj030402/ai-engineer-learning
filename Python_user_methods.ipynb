```python
# 1 Positional
  # Variable sequence matter when calling func
# 2 Keyword
  # when calling function must provide variable name and value , sequence not matter
# 3 Default
  # when we are not passing values from calling that time default valau take
# 4 Veriable Length args
  # i variable length positiional args - (*args) - return tuple
  # ii variable length keywork args - (** kwargs) - return dictinary
```

```python
# 1 Positional

def sum(a,b):
  print(a+b)
sum(10,20)
```

```python
# 2 Keyword

def detail(name,age):
  print(f"name : {name} , age : {age}")
detail(age=24,name="Yuvraj")
```

```python
# 3 Default
def detail(age , name = "Raj"):
  print(f"name : {name} , age : {age}")
detail(age=24)
detail(age=24, name = "yuvraj")
```

```python
# 4 Veriable Length args
  # i variable length positiional args
def info(*args):
  print(args)
info("Raj" , 23,"Pune")
```

```python
# 4 Veriable Length args
  # i variable length positiional args
def sum(*args):
    total = 0
    for n in args:
        total += n
    print(total)

sum(10, 20)
sum(10, 20, 30)
sum(10, 20, 30, 40)
```

```python
# 4 Veriable Length args
# ii variable length keywork args -
def info(**kwargs):
  print(kwargs)
info(name="raj",age = 24)
```

```python
  # i variable length positiional args - (*args) - return tuple
  # ii variable length keywork args - (** kwargs) - return dictinary
def info(*args , **kwargs):
   print(args)
   print(kwargs)
info("yuvraj",24 , address = "Pune")
```
