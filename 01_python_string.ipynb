```python
#2. Write a Python program to count vowels and consonants in a string.
#Input Programming
I = "Programming"
vowel = ['a','e','i','o','u']
v_Count = 0
c_Count = 0
for i in I.lower():
  if i in vowel:
    v_Count = v_Count + 1
  elif i.isalpha():
     c_Count = c_Count + 1
print(f"Vowel Count {v_Count}")
print(f'Consonants Count {c_Count}')
```

```python
#3 Write a Python program to count the frequency of each word in a sentence, ignoring letter case.
#Input Python is fun and python is powerful
#Expected Output python: 2, is: 2, fun: 1, and: 1, powerful: 1

I = 'Python is fun and python is powerful'
i_lower = I.lower()
d = {}
for i in i_lower.split(' '):
  d.update({i : i_lower.count(i)})
d
```

```python
#4. Write a Python program to find the longest word in a sentence.
# Input Python makes programming enjoyable
#Output Longest word = programming
#Length = 11

I = 'Python makes programming enjoyable'
max_length_text = 0
max_length_count = 0
for i in I.split(' '):
  if len(i) > max_length_count:
    max_length_count = len(i)
    max_length_text = i
print(f'Longest word :{max_length_text}')
print(f'Length :{max_length_count}')
```

```python
#5. Write a Python program to remove punctuation marks from a string.
#Input Hello, World! Python is fun.
#Expected Output Hello World Python is fun
I = 'Hello, World! Python is fun.'
s = ''
for i in I:
  if i.isalpha() or i.isspace() :
     s += i
s
```

```python
#6. Write a Python program to find all starting positions of a substring, including overlapping occurrences.
#String = aaaaa Substring = aa
#Positions = [0, 1, 2, 3]

s = 'aaaaa'
sb = 'aa'

for i in range(len(s) - len(sb) + 1):
    if s[i:i + len(sb)] == sb:
        print(i)
```

```python
#8. Write a Python program to compress consecutive repeated characters using character counts.
#Expected Output Compressed string = a3b2c4d1
Input =  'aaabbccccd'
output = ''
for i in Input:
  if i in output:
    continue
  count = 0
  count += Input.count(i)
  output += i + str(count)
print(output)

```
