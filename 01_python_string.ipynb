```python
#2. Write a Python program to count vowels and consonants in a string.
#Input Programming
I = "Programming"
vowel = ['a','e','i','o','u']
v_Count = 0
c_Count = 0
for i in I.lower():
  if i in vowel:
    v_Count = v_Count + 1
  elif i.isalpha():
     c_Count = c_Count + 1
print(f"Vowel Count {v_Count}")
print(f'Consonants Count {c_Count}')
```
