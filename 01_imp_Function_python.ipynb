```python
# Function	Use	Example
# len()	Count items or characters	len("Python") → 6
# type()	Check data type	type(10) → int
# print()	Display output	print("Hello")
# input()	Take user input	input("Enter name: ")
# range()	Generate a sequence of numbers	range(5)
# sum()	Add numbers	sum([1, 2, 3]) → 6
# max()	Find the largest value	max([2, 8, 4]) → 8
# min()	Find the smallest value	min([2, 8, 4]) → 2
# sorted()	Return sorted items	sorted([3, 1, 2]) → [1, 2, 3]
# abs()	Absolute value	abs(-10) → 10
# enumerate()	Get index and value	enumerate("cat")
# zip()	Combine sequences	zip([1, 2], ['a', 'b'])
# any()	Check whether any item is true	any([False, True]) → True
# all()	Check whether all items are true	all([True, True]) → True
```

```python
names = ['Rahul', 'Amit', 'Priya']
marks = [80, 90, 85]

for name, mark in zip(names, marks):
    print(name, mark)
```

```python
print(any([False, False, False]))  # False
print(any([True, False, False]))   # True
print(any([True, True, True]))     # True
```

```python
print(all([True, True, True]))    # True
print(all([True, False, True]))   # False
print(all([False, False, False])) # False
```
