```python
mark = 34

if mark >= 80:
  print("Grade : A")
elif mark >= 35 and mark <80:
  print("Grade : B")
else:
 print("Failed")
```

**Loops**

```python
#for - itetrate until last element
#while - iterate until cond not meet
```

```python
for i in range(5):
  print(i)
```

```python
age = 100
while age >= 18:
  print(f"Eligible age : {age}")
  age -= 1
```

Loop **control**

```python
#break
#continoue
#pass
```

```python
experince = [3,4,6,2,0]
for i in experince:
  if i >= 3:
    print("eligible")
  elif i > 2.8:
    pass
  elif i == 2:
    print("continue")
    continue
  elif i == 0:
    print("break")
    break
  else:
    pass
```

Function **control**

```python
# 1. return — Return a result
# return sends a result back to the place where the function was called and ends the function.
```

```python

def add(a, b):
    return a + b

result = add(10, 20)
print(result)
```

```python
# 2. yield — Return values one at a time
# yield produces a value and pauses the function. When the next value is requested, the function continues from where it paused.
```

```python

def numbers():
    yield 1
    yield 2
    yield 3

for n in numbers():
    print(n)
```
